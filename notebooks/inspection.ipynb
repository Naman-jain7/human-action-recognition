```python
FILE_PATH = "nturgb+d_skeletons/S001C001P001R001A001.skeleton"

with open(FILE_PATH, 'r') as f:
    lines = f.readlines()

lines[:20]
```

```python
def parse_skeleton(file_path):
    with open(file_path, 'r') as f:
        num_frames = int(f.readline())
        frame_lengths = []
        persons_per_frame = []

        for _ in range(num_frames):
            num_persons = int(f.readline())
            persons_per_frame.append(num_persons)

            for _ in range(num_persons):
                f.readline()  # body info
                num_joints = int(f.readline())
                frame_lengths.append(num_joints)

                for _ in range(num_joints):
                    f.readline()

    return num_frames, persons_per_frame, frame_lengths

print(parse_skeleton(file_path=FILE_PATH))
```

```python
import os

def get_label(filename):
    return int(filename.split('A')[-1].split('.')[0])

files = os.listdir("nturgb+d_skeletons")
labels = [get_label(f) for f in files]

print("Unique classes:", len(set(labels)))
```

```python
def read_first_frame(file_path):
    with open(FILE_PATH, 'r') as f:
        num_frames = int(f.readline())
        num_persons = int(f.readline())
        f.readline()  # body info
        num_joints = int(f.readline())

        joints = []
        for _ in range(num_joints):
            data = list(map(float, f.readline().split()))
            joints.append(data[:3])  # x, y, z

    return joints

print(read_first_frame(file_path=FILE_PATH)[:5])
```

```python
lengths = []

for f in files[:1000]:  # sample first
    n_frames, _, _ = parse_skeleton(os.path.join('nturgb+d_skeletons', f))
    lengths.append(n_frames)

print(min(lengths), max(lengths), sum(lengths)/len(lengths))
```

```python
def check_empty_frames(file_path):
    with open(file_path, 'r') as f:
        num_frames = int(f.readline())
        empty_frames = 0

        for _ in range(num_frames):
            num_persons = int(f.readline())
            if num_persons == 0:
                empty_frames += 1
            for _ in range(num_persons):
                f.readline()
                num_joints = int(f.readline())
                for _ in range(num_joints):
                    f.readline()

    return empty_frames

print(check_empty_frames(FILE_PATH))
```
